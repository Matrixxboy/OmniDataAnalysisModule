```python
# ── All Comprehension Types ───────────────────
# List
primes = [n for n in range(2,50) if all(n%i!=0 for i in range(2,int(n**0.5)+1))]
print("Primes:", primes)

# Dict
skills = ["Python","SQL","ML","Django"]
import random
scored = {s: random.randint(75,99) for s in skills}
print("Scored:", scored)

# Set — unique email domains
emails  = ["utsav@omni.com","rahul@omni.com","priya@tech.com"]
domains = {e.split("@")[1] for e in emails}
print("Domains:", domains)

# Generator expression
total = sum(x**2 for x in range(1,101))
print(f"Sum of squares 1-100: {total}")
```

```python
# ── Generator Function ───────────────────────
def paginate(items, page_size=3):
    for i in range(0, len(items), page_size):
        yield items[i:i+page_size]

jobs = ["Python Dev","ML Eng","Data Analyst","Backend Dev","AI Researcher","SQL Analyst"]
for pg, page in enumerate(paginate(jobs), 1):
    print(f"  Page {pg}: {page}")

# Infinite ID generator
def id_gen(prefix="OCH"):
    n = 1000
    while True:
        yield f"{prefix}-{n}"
        n += 1

gen = id_gen()
print([next(gen) for _ in range(5)])
```

```python
# ── functools ─────────────────────────────────
from functools import partial, lru_cache, reduce
import time

@lru_cache(maxsize=128)
def fib(n):
    if n < 2: return n
    return fib(n-1) + fib(n-2)

t0 = time.time()
print([fib(i) for i in range(30)])
print(f"Time: {(time.time()-t0)*1000:.2f}ms | {fib.cache_info()}")

# partial
apply_tax   = lambda amt, rate: amt * (1 + rate)
gst_18      = partial(apply_tax, rate=0.18)
print(f"₹100000 + 18% GST = ₹{gst_18(100000):,.0f}")
```
