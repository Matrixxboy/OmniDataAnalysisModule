## Regex Quick Reference

| Pattern | Matches |
|---------|---------|
| `.` | Any character |
| `\d` | Digit |
| `\w` | Word char |
| `\s` | Whitespace |
| `+` | 1 or more |
| `*` | 0 or more |
| `?` | Optional |
| `^` | Start |
| `$` | End |
| `[abc]` | Character class |


```python
import re

text = "Contact Utsav at utsav@omnicareerhub.com or +91-9876543210"

email = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)
phone = re.findall(r"\+?\d[\d-]{9,}", text)
print("Emails:", email)
print("Phones:", phone)

url = "https://github.com/Matrixxboy"
print("GitHub valid:", bool(re.match(r"^https://github\.com/[\w-]+", url)))
```

```python
# ── Resume Parser ────────────────────────────
resume = (
    'Name: Utsav Lankapati\n'
    'Email: utsav@omnicareerhub.com\n'
    'Phone: +91 98765 43210\n'
    'GitHub: https://github.com/Matrixxboy\n'
    'Portfolio: https://utsav-lankapati.onrender.com/\n'
    'Experience: 3.5 years\n'
    'Skills: Python, Machine Learning, SQL, Django, FastAPI\n'
)

patterns = {
    "Email"     : r"[\w.-]+@[\w.-]+\.\w+",
    "Phone"     : r"\+?[\d ]{10,}",
    "GitHub"    : r"https://github\.com/[\w-]+",
    "Experience": r"([\d.]+)\s*years?",
    "Portfolio" : r"https://[\w.-]+\.onrender\.com/?",
}

print("=== Parsed Resume ===")
for field, pat in patterns.items():
    m = re.search(pat, resume)
    print(f"  {field:<14}: {m.group(0).strip() if m else 'Not found'}")

skills_line = re.search(r"Skills:\s*(.+)", resume)
if skills_line:
    skills = [s.strip() for s in skills_line.group(1).split(',')]
    print(f"  {'Skills':<14}: {skills}")
```
