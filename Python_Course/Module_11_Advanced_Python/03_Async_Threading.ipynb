```python
import threading, time

def process_resume(name, delay):
    print(f"  🔄 Processing: {name}")
    time.sleep(delay)
    print(f"  ✅ Done: {name}")

resumes = [("Utsav Lankapati",0.5),("Priya Mehta",0.3),("Rahul Gupta",0.4),("Ankit Shah",0.2)]

print("=== Sequential ===")
t0 = time.time()
for name, delay in resumes:
    process_resume(name, delay)
print(f"Time: {time.time()-t0:.2f}s\n")

print("=== Threaded ===")
t0 = time.time()
threads = [threading.Thread(target=process_resume, args=(n,d)) for n,d in resumes]
for t in threads: t.start()
for t in threads: t.join()
print(f"Time: {time.time()-t0:.2f}s")
```

```python
import asyncio

async def fetch_jobs(company, delay):
    await asyncio.sleep(delay)
    return {"company": company, "jobs": ["Python Dev","ML Eng"]}

async def main():
    tasks = [
        fetch_jobs("Omni Career Hub", 0.3),
        fetch_jobs("TechCorp India",  0.5),
        fetch_jobs("DataFlow Labs",   0.2),
    ]
    results = await asyncio.gather(*tasks)
    for r in results:
        print(f"  {r['company']}: {r['jobs']}")

await main()
```
