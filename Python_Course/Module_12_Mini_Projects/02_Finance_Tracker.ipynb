```python
from datetime import datetime
from collections import defaultdict

class FinanceTracker:
    def __init__(self, owner="Utsav Lankapati"):
        self.owner = owner
        self.txns  = []

    def add(self, category, amount, txn_type="expense", note=""):
        self.txns.append({
            "date":     datetime.now().strftime("%Y-%m-%d"),
            "category": category,
            "amount":   amount,
            "type":     txn_type,
            "note":     note,
        })

    def summary(self):
        income  = sum(t["amount"] for t in self.txns if t["type"]=="income")
        expense = sum(t["amount"] for t in self.txns if t["type"]=="expense")
        savings = income - expense

        print(f"\n{'═'*52}")
        print(f"  Finance Tracker — {self.owner}")
        print(f"{'═'*52}")
        print(f"  💰 Income    : ₹{income:>12,.2f}")
        print(f"  💸 Expenses  : ₹{expense:>12,.2f}")
        print(f"  💹 Savings   : ₹{savings:>12,.2f}")
        print(f"  📊 Save Rate : {savings/income*100:>11.1f}%")
        print(f"{'─'*52}")

        cats = defaultdict(float)
        for t in self.txns:
            if t["type"]=="expense":
                cats[t["category"]] += t["amount"]

        print("  Expense Breakdown:")
        for cat, amt in sorted(cats.items(), key=lambda x: -x[1]):
            pct = amt/expense*100
            bar = "█"*int(pct/5)
            print(f"    {cat:<20} ₹{amt:>9,.0f}  {bar} {pct:.1f}%")
        print(f"{'═'*52}")

ft = FinanceTracker("Utsav Lankapati")
ft.add("Salary",     85000,"income", "Omni Career Hub")
ft.add("Freelance",  15000,"income", "Python tutoring")
ft.add("Portfolio",   5000,"income", "utsav-lankapati.onrender.com")
ft.add("Rent",       18000,"expense","Monthly")
ft.add("Food",        8000,"expense","Groceries + dining")
ft.add("Transport",   3000,"expense","Commute")
ft.add("Learning",    5000,"expense","Courses & books")
ft.add("Utilities",   2500,"expense","Internet + electricity")
ft.add("Entertainment",2000,"expense","Streaming")
ft.add("Health",      3500,"expense","Gym + medicines")
ft.summary()
```
