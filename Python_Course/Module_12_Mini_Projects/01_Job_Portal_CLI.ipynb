```python
import json
from datetime import datetime

class JobPortal:
    def __init__(self):
        self.jobs = [
            {"id":"J001","title":"Python Developer","skills":["Python","Django","SQL"],    "salary":80000, "apps":[]},
            {"id":"J002","title":"ML Engineer",      "skills":["Python","ML","TensorFlow"],"salary":110000,"apps":[]},
            {"id":"J003","title":"Data Analyst",     "skills":["SQL","Power BI","Excel"],  "salary":65000, "apps":[]},
            {"id":"J004","title":"Backend Developer","skills":["Python","FastAPI","Docker"],"salary":90000, "apps":[]},
            {"id":"J005","title":"Data Scientist",   "skills":["Python","ML","Statistics"],"salary":120000,"apps":[]},
        ]
        self.candidates = {
            "C001":{"name":"Utsav Lankapati","skills":["Python","ML","SQL","Django","TensorFlow"],"exp":3.5},
            "C002":{"name":"Priya Mehta",    "skills":["SQL","Power BI","Excel","Python"],       "exp":2.0},
            "C003":{"name":"Rahul Gupta",    "skills":["Python","FastAPI","Docker","SQL"],       "exp":1.5},
        }

    def list_jobs(self):
        print(f"\n{'='*65}")
        print("  OMNI CAREER HUB — Open Positions")
        print(f"{'='*65}")
        print(f"  {'ID':<6} {'Title':<25} {'Salary':>10}  Skills")
        print(f"  {'-'*60}")
        for j in self.jobs:
            print(f"  {j['id']:<6} {j['title']:<25} ₹{j['salary']:>9,}  {', '.join(j['skills'][:3])}")

    def match_jobs(self, cid):
        c = self.candidates.get(cid)
        if not c:
            return "❌ Candidate not found"
        print(f"\n🔍 Job Match for {c['name']}")
        matches = []
        for j in self.jobs:
            common = set(c["skills"]) & set(j["skills"])
            pct    = len(common) / len(j["skills"]) * 100
            matches.append((pct, j, common))
        for pct, j, common in sorted(matches, reverse=True, key=lambda x: x[0]):
            bar = "█"*int(pct/10) + "░"*(10-int(pct/10))
            print(f"  {j['title']:<25} [{bar}] {pct:5.1f}%")

    def apply(self, cid, jid):
        c = self.candidates.get(cid)
        j = next((x for x in self.jobs if x["id"]==jid), None)
        if not c or not j:
            return "❌ Invalid candidate or job ID"
        j["apps"].append({"name":c["name"],"at":datetime.now().strftime("%Y-%m-%d %H:%M")})
        return f"✅ {c['name']} applied for '{j['title']}'"

    def report(self):
        print("\n📊 Application Report")
        print("=" * 50)
        for j in self.jobs:
            n   = len(j["apps"])
            bar = "█" * n if n else "—"
            print(f"  {j['title']:<28} {bar} ({n})")

portal = JobPortal()
portal.list_jobs()
portal.match_jobs("C001")
portal.match_jobs("C002")
print()
print(portal.apply("C001","J002"))
print(portal.apply("C001","J001"))
print(portal.apply("C002","J003"))
print(portal.apply("C003","J004"))
portal.report()
```
