````python
# ── Basic Function ────────────────────────────
def greet(name, role="Learner"):
    return f"Hello {name}! Welcome as {role} at Omni Career Hub 👋"

print(greet("Utsav", "Instructor"))
print(greet("Rahul"))
````

````python
# ── *args and **kwargs ────────────────────────
def skill_report(*skills):
    print(f"Skills ({len(skills)}):")
    for s in skills:
        print(f"  • {s}")

skill_report("Python","ML","SQL","Django","Power BI")

def register(**info):
    print("Registration — Omni Career Hub")
    for k, v in info.items():
        print(f"  {k:<15}: {v}")

register(
    name="Utsav Lankapati",
    github="https://github.com/Matrixxboy",
    portfolio="https://utsav-lankapati.onrender.com/",
    experience=3.5
)
````

````python
# ── Lambda Functions ─────────────────────────
grade    = lambda s: "A" if s>=90 else "B" if s>=75 else "C"
gst      = lambda amt, rate=0.18: amt * (1 + rate)
name_fmt = lambda n: n.title().strip()

print(grade(88))
print(f"Price with GST: ₹{gst(1000):,.0f}")
print(name_fmt("  utsav lankapati  "))

# Lambda with sorted
employees = [
    {"name":"Utsav","salary":85000},
    {"name":"Priya","salary":65000},
    {"name":"Rahul","salary":72000},
]
for e in sorted(employees, key=lambda x: -x["salary"]):
    print(f"  {e['name']:<8} ₹{e['salary']:,}")
````

````python
# ── Recursion ─────────────────────────────────
def factorial(n):
    if n <= 1: return 1
    return n * factorial(n-1)

def fibonacci(n):
    if n < 2: return n
    return fibonacci(n-1) + fibonacci(n-2)

print("Factorials:", [factorial(i) for i in range(1,8)])
print("Fibonacci :", [fibonacci(i) for i in range(10)])
````
