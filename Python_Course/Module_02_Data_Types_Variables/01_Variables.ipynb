## 1. 📖 What is a Variable?
A **variable** is a named container that stores data.  Python is **dynamically typed** — no need to declare types.

```python
variable_name = value
```

| Rule | Example |
|------|---------|
| Start with letter or `_` | `name`, `_count` |
| Letters, digits, `_` only | `user_1` |
| Case-sensitive | `Name ≠ name` |
| No reserved keywords | ❌ `if = 5` |
