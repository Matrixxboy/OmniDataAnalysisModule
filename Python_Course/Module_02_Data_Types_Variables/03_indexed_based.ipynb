## str /String (Index-Based Perspective) :
A string is a built-in Python data structure used to store a sequence of characters. It is ordered and immutable, which means each character has a fixed position (index), and you cannot change it after creation.

* Ordered (each character has an index starting from 0)
* Supports positive and negative indexing
* Immutable (cannot modify characters directly)
* Supports slicing to extract parts of the string
* Allows duplicate characters

````python
# String example
text = "Python"

print(text)        # Python
````

```P   y   t   h   o   n```

```0   1   2   3   4   5```   → Positive Index

```-6  -5  -4  -3  -2  -1```   → Negative Index

````python
print(text[0])    # P
print(text[2])    # t
print(text[-6])   # n
````

````python
print(text[0:3])   # Pyt
print(text[2:5])   # tho
print(text[:4])    # Pyth
print(text[3:])    # hon
````

````python
# This will give error
# text[0] = "J"
````

### Array / List
Python provides multiple data structures for storing collections of values, among which Lists and Arrays are two commonly used options. While both support indexing, iteration and storage of multiple elements, they differ significantly in terms of memory usage, data type flexibility and performance.


### Array :
An array is a data structure that stores elements of the same data type in contiguous memory locations, making it efficient for numerical operations. Arrays require the array module to be used.
- Stores homogeneous data only
- Uses less memory than lists
- Faster for numerical and mathematical operations
- Better suited for large numeric datasets

````python
# Array Syntax
import array as arr
a = arr.array('i', [1, 2, 3])
print(type(a))
for i in a:
    print(i, end=" ")
````

### List : 
A list is a built-in Python data structure used to store an ordered collection of elements that can belong to different data types. Lists are dynamic, meaning their size can change during execution.
* Stores heterogeneous data like numbers, strings, objects, etc.
* Supports dynamic resizing (easy insertion and deletion)
* Allows duplicate values
* Supports indexing, slicing and nesting

````python
# List syntax
my_list = [1, 2, 3]
print(type(my_list))
for i in my_list:
    print(i, end=" ")
````

````python
# list can store the different datatypes values
# example : 
my_info = ["Utsav", 21, 175.5, True]

my_name  = my_info[0]
my_age = my_info[1]
my_height = my_info[2]
my_is_student = my_info[3]

print("my_name : ",my_name)
print("my_age : ",my_age)
print("my_height : ",my_height)
print("my_is_student : ",my_is_student)
````

### Tuple :

A tuple is a built-in Python data structure used to store an ordered collection of elements that can belong to different data types. Tuples are **immutable**, meaning once created, their elements cannot be changed.

* Stores heterogeneous data like numbers, strings, objects, etc.
* **Immutable** (no insertion, deletion, or modification after creation)
* Allows duplicate values
* Supports indexing, slicing, and nesting
* Faster than lists due to immutability
* Can be used as keys in dictionaries (because they are hashable, if elements are also hashable)


````python
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple)
print(type(my_tuple))
````

````python
print(my_tuple[0])
````

````python
# Predict the output before executing
# my_tuple[0] = 10
````

### Set :

A set is a built-in Python data structure used to store an **unordered collection of unique elements**. Sets are mutable, but they do not allow duplicate values.

* Stores heterogeneous data like numbers, strings, etc.
* **Unordered** (no indexing or slicing)
* **Does not allow duplicate values**
* Mutable (you can add or remove elements)
* Supports mathematical operations like union, intersection, difference
* Elements must be **hashable** (cannot store lists or other sets)


````python
# Tuple example
my_tuple = (1, "hello", 3.5)
print(my_tuple)       
````

````python
# Access element
print(my_tuple[1])  
````

````python
# ❌ This will give error (immutable)
# my_tuple[0] = 100
````

## Dictionary :

A dictionary is a built-in Python data structure used to store data in key–value pairs. It is unordered (in concept), mutable, and accessed using keys instead of index.

* Stores data in key : value format
* Keys must be unique and hashable (e.g., string, number, tuple)
* Values can be of any data type
* Mutable (you can add, update, delete elements)
* Access elements using keys, not index
* Faster lookup using keys

````python
# Dictionary example
student = {
    "name": "Utsav",
    "age": 20,
    "course": "Computer Engineering"
}

print(student)
````

````python
print(student["name"])    
print(student["age"])     
````

````python
student["age"] = 21          
student["city"] = "Surat"

print(student)
````

````python
# This will give error
print(student[0])
````
