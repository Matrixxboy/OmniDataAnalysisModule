```python
# ── for loop basics ──────────────────────────
skills = ["Python","SQL","ML","Power BI","Django"]

print("Skills at Omni Career Hub:")
for i, skill in enumerate(skills, 1):
    print(f"  {i}. {skill}")
```

```python
# ── zip() — parallel iteration ────────────────
students = ["Utsav","Priya","Rahul","Ankit"]
scores   = [95, 88, 82, 91]

print(f"{'Name':<12} {'Score':>6} {'Grade':>7}")
print("-" * 28)
for name, score in zip(students, scores):
    grade = "A" if score >= 90 else "B"
    print(f"{name:<12} {score:>6} {grade:>7}")
```

```python
# ── while loop ───────────────────────────────
attempts = 0
max_attempts = 3

while attempts < max_attempts:
    attempts += 1
    print(f"Login attempt {attempts}/3...")

print("Access granted!" if attempts < 3 else "Account locked!")

# break / continue
print("\nOdd numbers 1-10:")
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

```python
# ── Comprehensions ───────────────────────────
# List comprehension
squares = [x**2 for x in range(1, 11)]
print("Squares:", squares)

# Dict comprehension
skill_map = {s: len(s) for s in ["Python","SQL","ML","Django"]}
print("Skill → Length:", skill_map)

# Filtered comprehension
names  = ["Utsav","Priya","Rahul","Ankit","Sneha"]
scores2 = [95,88,72,91,85]
top = [n for n,s in zip(names,scores2) if s >= 90]
print("Top scorers:", top)

# Nested — flatten 2D list
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flat = [v for row in matrix for v in row]
print("Flattened:", flat)
```
