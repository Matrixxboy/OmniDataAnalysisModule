````python
# ── Grade Calculator ─────────────────────────
score = 87

if score >= 90:
    grade = "A — Excellent!"
elif score >= 75:
    grade = "B — Good!"
elif score >= 60:
    grade = "C — Average"
elif score >= 40:
    grade = "D — Below Average"
else:
    grade = "F — Needs Improvement"

print(f"Score: {score} | Grade: {grade}")
````

````python
# ── Job Application Filter — Omni Career Hub ──
applicant = {
    "name"     : "Riya Sharma",
    "experience": 3,
    "score"    : 88,
    "projects" : 5,
}

exp_ok  = applicant["experience"] >= 2
scr_ok  = applicant["score"] >= 75
proj_ok = applicant["projects"] >= 3

if exp_ok and scr_ok and proj_ok:
    status = "✅ SELECTED — Schedule Interview"
elif exp_ok and scr_ok:
    status = "⏳ WAITLISTED — Needs more projects"
elif scr_ok:
    status = "📋 REVIEW — Good score, low experience"
else:
    status = "❌ REJECTED"

print(f"Applicant : {applicant['name']}")
print(f"Status    : {status}")
````

````python
# ── Ternary Expression ───────────────────────
marks = 72
result = "Pass" if marks >= 40 else "Fail"
print(result)

# Match-Case (Python 3.10+)
day = "MON"
match day:
    case "SAT" | "SUN":
        print("Weekend!")
    case "MON"|"TUE"|"WED"|"THU"|"FRI":
        print("Weekday — Work at Omni Career Hub")
    case _:
        print("Invalid day")
````
