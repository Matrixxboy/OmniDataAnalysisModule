```python
# ── Numeric Types in Action ────────────────────────────
age   = 25
gpa   = 3.85
comp  = 3 + 4j
```

```python
print("age Data type : ",type(age))
print("gpa Data type : ",type(gpa))
print("complex Data type : ",type(comp) , "\nreal : ",comp.real ,"\nImg : ",comp.imag )
```

```python
print("Before")
print("Type cast gpa into float :" , gpa)
print("age Data type : ",type(gpa))
print("\nafter")
gpa = int(gpa)
print("Type cast age into float :" , gpa)
print("age Data type : ",type(gpa))
```
