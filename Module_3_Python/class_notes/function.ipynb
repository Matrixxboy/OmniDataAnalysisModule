```python
# *args example
def sum_numbers(*args):
    return sum(args)
```

```python
sum_numbers(1,2)
```

```python
sum_numbers(1,2,4,5,6,7,8,9,0)
```

```python
student = {
    "name": "Utsav", # 0
    "age": "wow",
    "course": "Computer Engineering",
    "status": "Active"
}

def show_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_profile(**student)
```

```python
# recursion function
def factorial(n):
    if n <= 1: return 1
    return n * factorial(n-1)

factorial(25)
```
