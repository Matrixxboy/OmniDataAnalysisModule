# Python `enumerate()` Function

The **`enumerate()`** function is a built-in Python function that allows you to loop through an iterable—such as a **list, tuple, or string**—while automatically keeping track of the **index** of each item.

Instead of manually creating and updating a counter variable, `enumerate()` gives you both the **index** and the **item** during each iteration.

---

## Syntax

```python
enumerate(iterable, start=0)
```

### Parameters

| Parameter            | Description                                                                |
| -------------------- | -------------------------------------------------------------------------- |
| `iterable`           | The collection you want to loop through, such as a list, tuple, or string. |
| `start` *(optional)* | The number from which the index counter should begin. Defaults to `0`.     |

---

## Why Use `enumerate()`?

Suppose you want to print a list of tasks along with their line numbers.

### ❌ Un-Pythonic Way: Manual Counter

You can create a separate variable to keep track of the index:

```python
tasks = ["Write code", "Review PR", "Deploy app"]

index = 0

for task in tasks:
    print(f"Task {index}: {task}")
    index += 1
```

This approach works, but it has some drawbacks:

* You have to create a separate counter.
* You must manually increment the counter.
* It's easy to forget `index += 1`.
* The code is more verbose than necessary.

---

### ✅ Clean, Pythonic Way: Using `enumerate()`

With `enumerate()`, Python automatically keeps track of the index:

```python
tasks = ["Write code", "Review PR", "Deploy app"]

for index, task in enumerate(tasks):
    print(f"Task {index}: {task}")
```

### Output

```text
Task 0: Write code
Task 1: Review PR
Task 2: Deploy app
```

The expression:

```python
for index, task in enumerate(tasks):
```

automatically provides two values during every iteration:

* `index` → The current item's position.
* `task` → The current item itself.

---

## Customizing the Start Index

By default, `enumerate()` starts counting from `0`.

However, if you're creating a numbered list for humans, you may want the numbering to start from `1`.

You can change the starting index using the `start` parameter:

```python
fruits = ["Apple", "Banana", "Cherry"]

for rank, fruit in enumerate(fruits, start=1):
    print(f"{rank}. {fruit}")
```

### Output

```text
1. Apple
2. Banana
3. Cherry
```

The following:

```python
enumerate(fruits, start=1)
```

means that the first item will have an index of `1` instead of `0`.

---

## How `enumerate()` Works Under the Hood

Calling `enumerate()` does **not** immediately create or print a list.

Instead, it returns an **`enumerate` object**, which is an **iterator**.

For example:

```python
colors = ["red", "green", "blue"]

result = enumerate(colors)

print(result)
```

The output will look similar to:

```text
<enumerate object at 0x...>
```

If you want to see all the index-item pairs at once, you can convert the `enumerate` object into a list using `list()`:

```python
colors = ["red", "green", "blue"]

print(list(enumerate(colors)))
```

### Output

```text
[(0, 'red'), (1, 'green'), (2, 'blue')]
```

Each element is a tuple containing:

```text
(index, item)
```

For example:

```text
(0, 'red')
(1, 'green')
(2, 'blue')
```

---

## Quick Summary

| Feature           | Description                                        |
| ----------------- | -------------------------------------------------- |
| **Purpose**       | Loop over items while keeping track of their index |
| **Default start** | `0`                                                |
| **Custom start**  | Use `start=`                                       |
| **Return value**  | An `enumerate` iterator                            |
| **Common usage**  | `for index, item in enumerate(iterable):`          |

### Recommended Pattern

Whenever you need both the **index** and the **value** while looping, prefer:

```python
for index, item in enumerate(items):
    print(index, item)
```

This is cleaner, safer, and more **Pythonic** than maintaining a manual counter.
