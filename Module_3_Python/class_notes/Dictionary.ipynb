# Python `zip()` Function

The **`zip()`** function is a built-in Python function used to **combine multiple iterables**—such as lists, tuples, or strings—into a single iterator of tuples.

It pairs elements based on their **position (index)**:

* First element → first element
* Second element → second element
* Third element → third element
* And so on

### 💡 Think of It Like a Zipper

Imagine a physical jacket zipper. It connects two sides together **element by element**.

Similarly, Python's `zip()` connects corresponding elements from multiple iterables.

---

## Syntax

```python
zip(iterable1, iterable2, ...)
```

You can pass **two or more iterables** to `zip()`.

---

## How `zip()` Works

Suppose you have a list of names and a list of scores:

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

# zip() returns a lazy iterator,
# so we convert it to a list to view the result
zipped_data = list(zip(names, scores))

print(zipped_data)
```

### Output

```text
[('Alice', 85), ('Bob', 92), ('Charlie', 78)]
```

Each pair contains:

```text
(name, score)
```

So the data is effectively combined like this:

| Name    | Score |
| ------- | ----: |
| Alice   |    85 |
| Bob     |    92 |
| Charlie |    78 |

---

# Key Features of `zip()`

## 1. Loop Through Multiple Lists Simultaneously

One of the most common uses of `zip()` is iterating through multiple lists at the same time.

Instead of manually managing index numbers, you can directly unpack each pair:

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(f"{name} scored {score}")
```

### Output

```text
Alice scored 85
Bob scored 92
Charlie scored 78
```

This is much cleaner than manually using indexes:

```python
for i in range(len(names)):
    print(f"{names[i]} scored {scores[i]}")
```

Using `zip()` makes the relationship between the two lists much easier to understand.

---

## 2. Handling Unequal Lengths — The Shortest Wins

When the iterables passed to `zip()` have different lengths, `zip()` stops as soon as the **shortest iterable is exhausted**.

For example:

```python
names = ["Alice", "Bob", "Charlie", "David"]  # 4 items
scores = [85, 92]                             # 2 items

print(list(zip(names, scores)))
```

### Output

```text
[('Alice', 85), ('Bob', 92)]
```

Notice that:

```text
"Charlie"
"David"
```

are not included because the `scores` list has no corresponding values.

### ⚠️ Important

By default:

> **`zip()` always stops at the shortest iterable.**

If you need to preserve the remaining elements, you can use `zip_longest()` from Python's built-in `itertools` module.

```python
from itertools import zip_longest

names = ["Alice", "Bob", "Charlie", "David"]
scores = [85, 92]

result = list(zip_longest(names, scores, fillvalue="N/A"))

print(result)
```

### Output

```text
[('Alice', 85), ('Bob', 92), ('Charlie', 'N/A'), ('David', 'N/A')]
```

---

## 3. Creating Dictionaries Quickly

`zip()` is also useful for combining a list of **keys** and a list of **values** into a dictionary.

For example:

```python
keys = ["username", "role"]
values = ["dev_core", "admin"]

user_dict = dict(zip(keys, values))

print(user_dict)
```

### Output

```text
{'username': 'dev_core', 'role': 'admin'}
```

Here's what happens:

```text
keys   → ["username", "role"]
values → ["dev_core", "admin"]
```

`zip()` pairs them:

```text
("username", "dev_core")
("role", "admin")
```

Then `dict()` converts those pairs into a dictionary:

```python
{
    "username": "dev_core",
    "role": "admin"
}
```

---

## 4. Unzipping Data

You can also perform the reverse operation—**splitting zipped pairs back into separate sequences**.

This is commonly called **unzipping**.

Consider this data:

```python
pairs = [
    ("Alice", 85),
    ("Bob", 92)
]
```

You can separate the names and scores using the unpacking `*` operator:

```python
names, scores = zip(*pairs)

print(names)
print(scores)
```

### Output

```text
('Alice', 'Bob')
(85, 92)
```

### How Does `zip(*pairs)` Work?

The `*` operator unpacks the list:

```python
zip(
    ("Alice", 85),
    ("Bob", 92)
)
```

Then `zip()` groups the corresponding elements:

```text
("Alice", "Bob")
(85, 92)
```

Finally, these two results are assigned to:

```python
names, scores
```

---

# `zip()` with More Than Two Iterables

`zip()` isn't limited to two iterables. You can combine three or more:

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 22]
cities = ["London", "New York", "Paris"]

for name, age, city in zip(names, ages, cities):
    print(f"{name} is {age} years old and lives in {city}.")
```

### Output

```text
Alice is 25 years old and lives in London.
Bob is 30 years old and lives in New York.
Charlie is 22 years old and lives in Paris.
```

Each iteration receives one item from **each iterable**.

---

# `zip()` Is Lazy

`zip()` returns a **lazy iterator**, rather than immediately creating a complete list.

For example:

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

result = zip(names, scores)

print(result)
```

The output will look similar to:

```text
<zip object at 0x...>
```

If you want to see all the values, convert it to a list:

```python
print(list(result))
```

Output:

```text
[('Alice', 85), ('Bob', 92), ('Charlie', 78)]
```

This lazy behavior allows `zip()` to work efficiently with large or even infinite iterables.

---

# Common Use Cases

`zip()` is particularly useful when you need to:

* Iterate over multiple lists simultaneously.
* Combine related data from separate sequences.
* Create dictionaries from keys and values.
* Pair records together.
* Transform tabular data.
* Unpack previously zipped data.

---

# Quick Summary

| Feature                 | Description                                    |
| ----------------------- | ---------------------------------------------- |
| **Purpose**             | Combines multiple iterables element by element |
| **Return value**        | A lazy `zip` iterator                          |
| **Pairing**             | Elements are paired by their position          |
| **Unequal lengths**     | Stops at the shortest iterable                 |
| **Multiple iterables**  | Supports two or more iterables                 |
| **Dictionary creation** | `dict(zip(keys, values))`                      |
| **Unzipping**           | `zip(*pairs)`                                  |
| **Keep extra elements** | Use `itertools.zip_longest()`                  |

---

## Recommended Pattern

Whenever you need to process corresponding elements from multiple iterables, `zip()` is usually the cleanest approach:

```python
for item1, item2 in zip(iterable1, iterable2):
    # Process both items
    ...
```

Instead of manually managing indexes:

```python
for i in range(len(iterable1)):
    item1 = iterable1[i]
    item2 = iterable2[i]
```

### ⭐ Key Takeaway

> **Use `zip()` when you want to work with corresponding elements from multiple iterables at the same time.**

It makes your code **shorter, cleaner, and more Pythonic**.
