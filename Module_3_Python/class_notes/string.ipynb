```python
name = "Utsav Lankapati"
url  = "https://github.com/Matrixxboy"


# ✂️ Slicing (Extracting Substrings)

print("\n--- Slicing ---") #error
print("First 5 chars:", name[:5])       

print("URL from index 8 onwards:  ", url[8:])       
print("Negative indexing (last 10):", url[-10:])     
print("Step (every 2nd char):     ", name[::2])     
```
