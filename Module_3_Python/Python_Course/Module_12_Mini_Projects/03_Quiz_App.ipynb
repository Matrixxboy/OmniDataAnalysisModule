```python
import random, time

class QuizApp:
    def __init__(self):
        self.questions = [
            {"q":"What is the output of: bool(0)?","opts":["True","False","None","Error"],"ans":1},
            {"q":"Which method adds to end of list?","opts":["insert()","add()","append()","push()"],"ans":2},
            {"q":"What does // do in Python?","opts":["Division","Modulus","Floor division","Power"],"ans":2},
            {"q":"Which is immutable?","opts":["List","Dict","Set","Tuple"],"ans":3},
            {"q":"Output of: len({1,2,2,3})?","opts":["4","3","2","Error"],"ans":1},
            {"q":"What is a lambda?","opts":["A loop","An anonymous function","A class","A module"],"ans":1},
            {"q":"Which opens a file safely?","opts":["open()","with open()","file()","read()"],"ans":1},
            {"q":"What does *args do?","opts":["Keyword args","Positional args","Both","Neither"],"ans":1},
        ]

    def run(self, name="Learner"):
        print(f"\n{'='*55}")
        print(f"  OMNI CAREER HUB — Python Quiz")
        print(f"  Player: {name}")
        print(f"{'='*55}")

        qs     = random.sample(self.questions, min(5, len(self.questions)))
        score  = 0
        start  = time.time()

        for i, q in enumerate(qs, 1):
            print(f"\nQ{i}: {q['q']}")
            for j, opt in enumerate(q["opts"], 1):
                print(f"  {j}. {opt}")
            try:
                ans = int(input("Your answer (1-4): "))
            except Exception:
                ans = 0
            if ans == q["ans"]:
                print("  ✅ Correct!")
                score += 1
            else:
                print(f"  ❌ Wrong. Answer: {q['ans']}. {q['opts'][q['ans']-1]}")

        elapsed = time.time() - start
        pct     = score / len(qs) * 100
        grade   = "A" if pct>=80 else "B" if pct>=60 else "C"

        print(f"\n{'='*55}")
        print(f"  Quiz Complete!")
        print(f"  Score : {score}/{len(qs)} ({pct:.0f}%) | Grade: {grade}")
        print(f"  Time  : {elapsed:.1f}s")
        print(f"  Status: {'🏆 Excellent!' if pct>=80 else '👍 Good!' if pct>=60 else '📚 Keep practicing!'}")
        print(f"{'='*55}")

# Simulate (auto-answer for demo)
quiz = QuizApp()
print("Quiz App ready! Call quiz.run('Your Name') to start.")
print("(In interactive mode you can play the quiz)")

# Demo summary
print("\nSample Quiz Summary:")
print("  Score : 4/5 (80%) | Grade: A")
print("  Time  : 42.3s")
print("  Status: 🏆 Excellent!")
```
