## 1. 📖 What is Python?
Python is a **high-level, interpreted, general-purpose** programming language created by **Guido van Rossum** (1991).
Python emphasizes:
- **Readability** — clean, English-like syntax
- **Simplicity** — fewer lines of code
- **Versatility** — web dev, data science, AI, automation

<div class='note'>💡 Python's name comes from <em>Monty Python's Flying Circus</em>, not the snake!</div>

---

## 3. 🔧 Where is Python Used?
- **Web Development** — Django, Flask, FastAPI
- **Data Science & Analytics** — Pandas, NumPy, Matplotlib
- **Machine Learning / AI** — TensorFlow, PyTorch, scikit-learn
- **Automation & Scripting** — Selenium, PyAutoGUI
- **Finance** — Algorithmic trading, risk analysis
- **Cybersecurity** — Penetration testing tools
