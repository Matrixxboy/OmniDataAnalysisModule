
## 2. 🖨️ The `print()` Function

```python
print(value, ..., sep=' ', end='\n')
```

| Parameter | Default | Description |
|-----------|---------|-------------|
| `value` | required | What to print |
| `sep` | `' '` | Separator between values |
| `end` | `'\n'` | End character |


````python
# Multiple values with custom separator
print("Omni", "Career", "Hub", sep="-")

# Custom end
print("Loading", end="...")
print("Done! ✅")

# Formatted print
name = "Utsav Lankapati"
print(f"Author  : {name}")
print(f"GitHub  : https://github.com/Matrixxboy")
print(f"Portfolio: https://utsav-lankapati.onrender.com/")
````
