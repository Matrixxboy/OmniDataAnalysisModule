```python
# ── Closures ──────────────────────────────────
def make_multiplier(factor):
    def multiply(n):
        return n * factor
    return multiply

double   = make_multiplier(2)
triple   = make_multiplier(3)
gst_calc = make_multiplier(1.18)

print(double(50))             # 100
print(triple(50))             # 150
print(f"₹{gst_calc(1000):.2f}")  # ₹1180.00
```

```python
# ── Decorators ────────────────────────────────
import time

def timer(func):
    def wrapper(*args, **kwargs):
        t0 = time.time()
        result = func(*args, **kwargs)
        print(f"⏱ {func.__name__} → {(time.time()-t0)*1000:.2f}ms")
        return result
    return wrapper

def logger(func):
    def wrapper(*args, **kwargs):
        print(f"📋 {func.__name__}({args})")
        result = func(*args, **kwargs)
        print(f"✅ returned: {result}")
        return result
    return wrapper

@timer
@logger
def skill_match(candidate_skills, job_skills):
    match = len(set(candidate_skills) & set(job_skills))
    return round(match / len(job_skills) * 100, 1)

utsav = ["Python","ML","SQL","Django","Power BI"]
job   = ["Python","SQL","ML","React"]
skill_match(utsav, job)
```

```python
# ── Generators ────────────────────────────────
def job_feed(company, n):
    roles = ["Data Scientist","ML Engineer","Python Dev","Data Analyst","AI Researcher"]
    for i in range(n):
        yield {"id": i+1, "company": company, "role": roles[i % len(roles)]}

for job in job_feed("Omni Career Hub", 5):
    print(f"  [{job['id']}] {job['company']} — {job['role']}")

# Generator for infinite IDs
def id_gen(prefix="OCH"):
    n = 1000
    while True:
        yield f"{prefix}-{n}"
        n += 1

gen = id_gen()
print([next(gen) for _ in range(5)])
```

```python
# ── map / filter / reduce ─────────────────────
from functools import reduce

salaries = [45000,62000,38000,75000,52000,91000]

after_tax = list(map(lambda s: s*0.7, salaries))
senior    = list(filter(lambda s: s>=60000, salaries))
total     = reduce(lambda a,b: a+b, salaries)

print("After tax:", [f"₹{s:,.0f}" for s in after_tax])
print("Senior >=60k:", senior)
print(f"Total payroll: ₹{total:,}")
```
