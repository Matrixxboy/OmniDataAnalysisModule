```python
# ── datetime ─────────────────────────────────
from datetime import datetime, date, timedelta

now   = datetime.now()
today = date.today()
print(f"Now    : {now.strftime('%d %B %Y %H:%M')}")
print(f"Today  : {today}")

deadline = today + timedelta(days=30)
print(f"Deadline: {deadline}")

# Course duration
start = date(2025, 1, 1)
end   = date(2025, 6, 30)
print(f"Duration: {(end-start).days} days")
```

```python
# ── random ───────────────────────────────────
import random

skills = ["Python","SQL","ML","Power BI","Django","FastAPI"]
print("Random skill :", random.choice(skills))
print("Random 3     :", random.sample(skills, 3))
print("Random int   :", random.randint(50, 100))

questions = ["What is OOP?", "Explain decorators", "What is a lambda?"]
random.shuffle(questions)
for i, q in enumerate(questions, 1):
    print(f"  Q{i}: {q}")
```

```python
# ── collections ──────────────────────────────
from collections import Counter, defaultdict, deque

apps = ["Python Dev","ML Eng","Python Dev","Data Analyst","ML Eng","Python Dev"]
print("Applications Counter:")
for role, n in Counter(apps).most_common():
    print(f"  {role:<20} {'█'*n} ({n})")

# defaultdict
skill_users = defaultdict(list)
for skill, user in [("Python","Utsav"),("SQL","Priya"),("Python","Rahul")]:
    skill_users[skill].append(user)
print("Skill→Users:", dict(skill_users))

# deque (fixed-size queue)
q = deque(maxlen=3)
for item in ["Job1","Job2","Job3","Job4"]:
    q.append(item)
print("Recent 3:", list(q))
```

```python
# ── math & string ────────────────────────────
import math, string

print(f"π = {math.pi:.6f}")
print(f"e = {math.e:.6f}")
print(f"sqrt(144) = {math.sqrt(144)}")
print(f"ceil(3.2) = {math.ceil(3.2)}")
print(f"7! = {math.factorial(7)}")

print("\nLetters:", string.ascii_letters[:12] + "...")
print("Digits :", string.digits)
print("Punct  :", string.punctuation[:10] + "...")
```
