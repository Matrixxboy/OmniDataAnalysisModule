## Exception Handling Syntax

```python
try:
    # risky code
except ExceptionType as e:
    # handle
else:
    # if no error
finally:
    # always runs
```

| Exception | Cause |
|-----------|-------|
| `ValueError` | Wrong value |
| `TypeError` | Wrong type |
| `IndexError` | Out of range |
| `KeyError` | Key not found |
| `ZeroDivisionError` | Divide by 0 |
| `FileNotFoundError` | File missing |


````python
# ── Basic try/except ──────────────────────────
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("❌ Cannot divide by zero!")
        return None
    else:
        return result
    finally:
        print(f"  (attempted {a}/{b})")

print(safe_divide(100, 4))
print(safe_divide(50, 0))
````

````python
# ── Multiple exception types ──────────────────
def process_score(data, idx):
    try:
        score = int(data[idx])
        if not 0 <= score <= 100:
            raise ValueError(f"Score {score} out of range")
        return f"Score: {score} | Grade: {'A' if score>=90 else 'B' if score>=75 else 'C'}"
    except IndexError:
        return "❌ Index out of range"
    except ValueError as e:
        return f"❌ {e}"
    except TypeError:
        return "❌ Non-numeric data"

data = ["95","82","abc","150","88"]
for i in range(7):
    print(f"  [{i}] → {process_score(data, i)}")
````

````python
# ── Custom Exceptions ────────────────────────
class OmniError(Exception):
    pass

class SkillGapError(OmniError):
    def __init__(self, missing):
        super().__init__(f"Missing skills: {', '.join(missing)}")
        self.missing = missing

class LowExperienceError(OmniError):
    def __init__(self, name, got, required):
        super().__init__(f"{name}: needs {required}yrs, has {got}yrs")

def validate(name, skills, exp):
    required = {"Python", "SQL"}
    missing  = required - set(skills)
    if missing:
        raise SkillGapError(missing)
    if exp < 1:
        raise LowExperienceError(name, exp, 1)
    return f"✅ {name} — Application Valid!"

for name, skills, exp in [
    ("Utsav", ["Python","SQL","ML"], 3.5),
    ("New Grad", ["Java"], 0),
    ("Part Timer", ["Python","SQL"], 0.5),
]:
    try:
        print(validate(name, skills, exp))
    except (SkillGapError, LowExperienceError) as e:
        print(f"❌ {e}")
````
