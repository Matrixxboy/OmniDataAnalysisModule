````python
# ── Creation & Access ────────────────────────
skills = ["Python","SQL","ML","Power BI","Django"]

print(skills[0])    # Python
print(skills[-1])   # Django
print(skills[1:3])  # ['SQL', 'ML']
print(len(skills))  # 5

# ── Methods ──────────────────────────────────
skills.append("FastAPI")
skills.insert(2, "Statistics")
skills.remove("Power BI")
popped = skills.pop()
print("After ops:", skills)
print("Popped:", popped)

skills.sort()
print("Sorted:", skills)
````
