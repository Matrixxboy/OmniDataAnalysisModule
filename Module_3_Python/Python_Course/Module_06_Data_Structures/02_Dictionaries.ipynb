```python
# ── Dictionary Basics ────────────────────────
profile = {
    "name"        : "Utsav Lankapati",
    "role"        : "Python Instructor",
    "organization": "Omni Career Hub",
    "github"      : "https://github.com/Matrixxboy",
    "portfolio"   : "https://utsav-lankapati.onrender.com/",
    "skills"      : ["Python","ML","Django","SQL"],
    "experience"  : 3.5,
}

# Access
print(profile["name"])
print(profile.get("email", "N/A"))   # safe

# Add / Update
profile["city"] = "Surat, India"
profile["experience"] = 4.0

# Iterate
for k, v in profile.items():
    print(f"  {k:<14}: {v}")
```

```python
# ── Dict Comprehension & Nested Dicts ─────────
skills  = ["Python","SQL","ML","Django"]
ratings = [95, 88, 91, 87]

skill_map = {s: r for s, r in zip(skills, ratings)}
top       = {k: v for k, v in skill_map.items() if v >= 90}
print("All:", skill_map)
print("Top:", top)

# Nested dict — Employee DB
employees = {
    "E001": {"name":"Utsav Lankapati","dept":"Instruction","salary":85000},
    "E002": {"name":"Priya Mehta",    "dept":"Analytics",  "salary":72000},
    "E003": {"name":"Rahul Gupta",    "dept":"Engineering","salary":78000},
}

print(f"\n{'ID':<6} {'Name':<20} {'Dept':<15} {'Salary':>10}")
print("-" * 55)
for eid, emp in employees.items():
    print(f"{eid:<6} {emp['name']:<20} {emp['dept']:<15} ₹{emp['salary']:>9,}")
```
