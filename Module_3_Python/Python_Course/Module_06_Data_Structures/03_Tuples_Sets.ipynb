```python
# ── SETS — Job Skill Matching ─────────────────
utsav_skills = {"Python","ML","SQL","Django","Power BI"}
job_required = {"Python","SQL","React","AWS","ML"}

common   = utsav_skills & job_required
missing  = job_required - utsav_skills
all_skls = utsav_skills | job_required

print(f"Candidate skills : {sorted(utsav_skills)}")
print(f"Job requirements : {sorted(job_required)}")
print(f"Matched skills   : {sorted(common)}")
print(f"Missing skills   : {sorted(missing)}")
print(f"Match Score      : {len(common)/len(job_required)*100:.1f}%")
```
