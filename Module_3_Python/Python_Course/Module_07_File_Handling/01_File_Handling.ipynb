## File Modes

| Mode | Description |
|------|-------------|
| `'r'` | Read (default) |
| `'w'` | Write (overwrites) |
| `'a'` | Append |
| `'x'` | Create (error if exists) |
| `'b'` | Binary suffix |

<div class='tip'>✅ Always use <code>with open()</code> — auto-closes the file!</div>


```python
import os

# ── Write ─────────────────────────────────────
with open("students.txt", "w") as f:
    f.write("Omni Career Hub — Student Roster\n")
    f.write("=" * 40 + "\n")
    for line in ["Utsav | Python | 95", "Priya | SQL | 88", "Rahul | ML | 82"]:
        f.write(line + "\n")
print("Written!")

# ── Read ──────────────────────────────────────
with open("students.txt") as f:
    print(f.read())

# ── Append ────────────────────────────────────
with open("students.txt", "a") as f:
    f.write("Ankit | Power BI | 91\n")
```

```python
import json

# ── JSON ──────────────────────────────────────
data = {
    "name"     : "Utsav Lankapati",
    "github"   : "https://github.com/Matrixxboy",
    "portfolio": "https://utsav-lankapati.onrender.com/",
    "org"      : "Omni Career Hub",
    "skills"   : ["Python","ML","Django"],
    "verified" : True
}

with open("profile.json","w") as f:
    json.dump(data, f, indent=4)

with open("profile.json") as f:
    loaded = json.load(f)

print(loaded["name"])
print(loaded["skills"])
```

```python
import csv

headers = ["Name","Course","Score","Grade"]
rows = [
    ["Utsav Lankapati","Python",95,"A"],
    ["Priya Mehta","SQL",88,"B"],
    ["Rahul Gupta","ML",82,"B"],
    ["Ankit Shah","Power BI",91,"A"],
]

with open("scores.csv","w",newline="") as f:
    csv.writer(f).writerow(headers)
    csv.writer(f).writerows(rows)

with open("scores.csv") as f:
    for row in csv.DictReader(f):
        print(f"{row['Name']:<20} | {row['Course']:<10} | {row['Score']} | {row['Grade']}")

# Cleanup
for fn in ["students.txt","profile.json","scores.csv"]:
    if os.path.exists(fn): os.remove(fn)
print("Cleanup done!")
```
