## Arithmetic Operators

| Op | Symbol | `5 op 2` |
|----|--------|----------|
| Add | `+` | `7` |
| Sub | `-` | `3` |
| Mul | `*` | `10` |
| Div | `/` | `2.5` |
| Floor Div | `//` | `2` |
| Modulus | `%` | `1` |
| Power | `**` | `25` |


```python
# ── Salary Calculator — Omni Career Hub ───────
base     = 60000
bonus_p  = 15
tax_rate = 0.30

bonus  = base * (bonus_p / 100)
gross  = base + bonus
tax    = gross * tax_rate
net    = gross - tax

print("=" * 42)
print("  OMNI CAREER HUB Salary Breakdown")
print("=" * 42)
print(f"  Base Salary : ₹{base:>10,.2f}")
print(f"  Bonus (15%) : ₹{bonus:>10,.2f}")
print(f"  Gross       : ₹{gross:>10,.2f}")
print(f"  Tax  (30%)  : ₹{tax:>10,.2f}")
print(f"  Net Salary  : ₹{net:>10,.2f}")
print("=" * 42)

# Floor div & modulus
weeks, days = divmod(365, 7)
print(f"\n365 days = {weeks}w {days}d")
```


## Comparison & Logical Operators

| Op | Meaning |
|----|---------|
| `==` `!=` | Equal / Not equal |
| `>` `<` `>=` `<=` | Comparison |
| `and` | Both True |
| `or` | At least one True |
| `not` | Invert |


```python
# ── Candidate Shortlisting ─────────────────
exp    = 3.5
score  = 88
background_ok = True

exp_ok   = exp   >= 2
score_ok = score >= 75

shortlisted = exp_ok and score_ok
print(f"Shortlisted : {shortlisted}")
print(f"70<=score<90: {70 <= score < 90}")  # chained

# Assignment operators
salary = 60000
salary += 5000
salary *= 1.10
print(f"Updated salary: ₹{salary:,.0f}")

# Identity & Membership
skills = ["Python","ML","SQL"]
print("Python" in skills)
print("Java" not in skills)
a = skills
print(a is skills)   # True — same object
```
