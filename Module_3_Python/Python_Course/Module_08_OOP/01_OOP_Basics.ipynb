## OOP Core Concepts

| Concept | Description |
|---------|-------------|
| Class | Blueprint / template |
| Object | Instance of class |
| Attribute | Data in object |
| Method | Function in class |
| `__init__` | Constructor |
| `self` | Current instance |

## 4 Pillars
1. **Encapsulation** — data + methods bundled
2. **Inheritance** — child inherits parent
3. **Polymorphism** — same name, different behavior
4. **Abstraction** — hide implementation details


```python
class Candidate:
    company = "Omni Career Hub"   # class variable

    def __init__(self, name, skills, experience):
        self.name       = name
        self.skills     = skills
        self.experience = experience
        self.applied    = []

    def apply(self, role):
        self.applied.append(role)
        return f"{self.name} applied for '{role}'"

    def match(self, required):
        pct = len(set(self.skills) & set(required)) / len(required) * 100
        return round(pct, 1)

    def __str__(self):
        return f"Candidate({self.name}, exp={self.experience}yrs)"

utsav = Candidate("Utsav Lankapati",["Python","ML","SQL","Django"],3.5)
priya = Candidate("Priya Mehta",    ["SQL","Power BI","Excel"],    2.0)

print(utsav)
print(utsav.apply("ML Engineer"))
print(utsav.apply("Python Developer"))
print(f"Match: {utsav.match(['Python','ML','React','SQL'])}%")
```

```python
# ── Inheritance & Polymorphism ────────────────
class Person:
    def __init__(self, name, email):
        self.name = name
        self.email = email
    def introduce(self):
        return f"Hi, I'm {self.name}"

class Instructor(Person):
    def __init__(self, name, email, courses, github):
        super().__init__(name, email)
        self.courses = courses
        self.github  = github
    def introduce(self):
        return f"{super().introduce()}, Instructor at Omni Career Hub"

class Student(Person):
    def __init__(self, name, email, course):
        super().__init__(name, email)
        self.course   = course
        self.progress = 0
    def introduce(self):
        return f"{super().introduce()}, studying '{self.course}'"
    def advance(self, pct):
        self.progress = min(100, self.progress + pct)

utsav = Instructor("Utsav Lankapati","utsav@omni.com",
                   ["Python","ML"],"https://github.com/Matrixxboy")
rahul = Student("Rahul Gupta","rahul@example.com","Python Basics")

for p in [utsav, rahul]:
    print(p.introduce())

rahul.advance(40)
print(f"\n{rahul.name} progress: {rahul.progress}%")
```

```python
# ── Encapsulation with @property ──────────────
class SalaryAccount:
    def __init__(self, employee, salary):
        self.employee = employee
        self.__salary = salary     # private

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, val):
        if val < 0:
            raise ValueError("Salary cannot be negative")
        self.__salary = val

    @property
    def annual(self):
        return self.__salary * 12

acc = SalaryAccount("Utsav Lankapati", 85000)
print(f"Monthly : ₹{acc.salary:,}")
print(f"Annual  : ₹{acc.annual:,}")
acc.salary = 95000
print(f"Updated : ₹{acc.salary:,}")
```
