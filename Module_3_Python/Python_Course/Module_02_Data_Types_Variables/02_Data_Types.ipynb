## 1️⃣ Numeric Types: `int`, `float`, `complex` 🔢
Python handles massive numbers with ease and comes with built-in math capabilities.

```python
# ── Numeric Types in Action ────────────────────────────
age   = 25
gpa   = 3.85
comp  = 3 + 4j

print("Type of age:", type(age), "| Value:", age)
print("Type of gpa:", type(gpa), "| Value:", gpa)
print("Type of comp:", type(comp), "| Real:", comp.real, "| Imaginary:", comp.imag)

# 🧮 Mind-Blowing Python Math (No Limits!)
massive_int = 10 ** 50 # 10 to the power of 50
print("Massive Integer:", massive_int)

# 🔄 Type Conversions (Casting)
print("float(age) :", float(age))    # 25 -> 25.0
print("int(gpa)   :", int(gpa))      # 3.85 -> 3 (Notice it truncates, doesn't round)
print("str(age)   :", str(age))      # 25 -> '25'
print("round(gpa) :", round(gpa, 1)) # Rounds 3.85 to 3.9
```

## 2️⃣ Strings: Text Manipulation Magic 🔤
Strings in Python are arrays of bytes representing Unicode characters. They are incredibly flexible.

```python
# ── String Operations ──────────────────────────────────
name = "Utsav Lankapati"
url  = "https://github.com/Matrixxboy"

print("Original:     ", name)
print("Upper case:   ", name.upper())
print("Lower case:   ", name.lower())
print("Title case:   ", "omni career hub".title())
print("Replace:      ", name.replace("Lankapati", "L."))
print("Length:       ", len(name))

# ✂️ Slicing (Extracting Substrings)
print("\n--- Slicing ---")
print("First 5 chars:             ", name[:5])       # Utsav
print("URL from index 8 onwards:  ", url[8:])        # github.com/Matrixxboy
print("Negative indexing (last 10):", url[-10:])     # Matrixxboy
print("Step (every 2nd char):     ", name[::2])      # UsvLnaai

# 💬 f-strings (String Interpolation - The Pythonic Way!)
org  = "Omni Career Hub"
exp  = 3.5
print("\n--- f-strings ---")
print(f"Hi, I am {name}. I work at {org} and have {exp} years of experience.")

# 📑 Multiline Strings
bio = f"""
Name     : {name}
GitHub   : {url}
Portfolio: https://utsav-lankapati.onrender.com/
"""
print(bio)
```

## 3️⃣ Booleans & None: The Logic Center 🤖
Booleans represent `True` or `False`. Under the hood, Python treats `True` as `1` and `False` as `0`.
`None` is a special constant representing the absence of a value or a null value.

```python
# ── Boolean & None ───────────────────────────
is_verified   = True
has_portfolio = True
is_student    = False

print("Boolean as Int (True): ", int(True))
print("Boolean as Int (False):", int(False))

# 🤔 Truthy & Falsy Concepts
# In Python, empty collections, 0, and None evaluate to False.
print("\n--- Truthy & Falsy ---")
print("bool(0)    ->", bool(0))      # False
print("bool('')   ->", bool(""))     # False
print("bool([])   ->", bool([]))     # False
print("bool(42)   ->", bool(42))     # True
print("bool('AI') ->", bool("AI"))   # True

# 👻 The 'None' Type
resume = None
if resume is None:
    print("\nPlease upload your resume! (Value is None)")

# ⚡ Logical Operators: and, or, not
profile_ready = is_verified and has_portfolio
needs_setup = not is_verified
print(f"Profile ready: {profile_ready}")
print(f"Needs setup:   {needs_setup}")
```
